Exercise 5: Write a program which prompts the user for a Celsius temperature, convert the temperature to Fahrenheit, and print out the converted temperature.

````python
def celsius_to_f(c_temp):
    f_temp = c_temp * 9/5 + 32
    return f_temp

user_input = input("What temperature do you want to convert from C?")

try:
    user_input = float(user_input)
    print(f"{user_input} in celsius is {celsius_to_f(user_input)} in Fahrenheit")
except ValueError:
    print("Bad input. Enter a number")
````

3.1

Rewrite your pay computation to give the employee 1.5 times the hourly rate for hours worked above 40 hours.

````python
def calculate_pay(hours, rate):
    if hours > 40:
        overtime = hours - 40
        overtime_pay = rate * 1.5 * overtime
        normal_pay = rate * 40
        total_pay = normal_pay + overtime_pay
    else:
        normal_pay = rate * hours
        total_pay = normal_pay
    return total_pay
````

````python
calculate_pay(20, 10)
````

````python
print(calculate_pay(40,10))
print(calculate_pay(rate=10, hours=50))
````

Write a program to prompt for a score between 0.0 and 1.0. If the score is out of range, print an error message. If the score is between 0.0 and 1.0, print a grade using the following table:


````python
score = input("Enter the score:")
score = float(score)

if score < 0 or score > 1:
    print("Enter a score between 0 and 1")
elif score > .9:
    grade = "A" + ". Great work!"
elif score > .8:
    grade = "B"
elif score > .7:
    grade = "C"
elif score > .6:
    grade = "D"
else:
    grade = "F"

print(f"You received a {grade}")
````
