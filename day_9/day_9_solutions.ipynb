We are going to get the hearings from Congress; as many as we can.

Here's an example URL:
https://api.congress.gov/v3/hearing?format=json&offset=0&limit=250

```python
import requests
import congress_auth
```

```python
import csv

def get_hearings(offset=0):
    base_url = 'https://api.congress.gov/v3/hearing' 
    params = {'format': 'json',
              'offset': offset,
              'limit': 250,
              'api_key': congress_auth.API_KEY}
    
    urls = []
    while True:
        r = requests.get(url=base_url, params=params)
        hearings = r.json()
        try:
            hearings['hearings']
        except KeyError:
            print(f'Error at offset {params["offset"]}')
            params['offset'] = params['offset'] + 1
            print(f'Trying again with offset {params["offset"]}')
            continue
        for hearing in hearings['hearings']:
            urls.append(hearing['url'])
        if 'next' in hearings['pagination']:
            params['offset'] = params['offset'] + 250
            print(f'Getting set {params["offset"]}')
        else:
            return urls
```

```python
urls = get_hearings(offset=0)

with open('hearing_urls.csv', 'w') as f:
    writer = csv.writer(f)
    for url in urls:
        writer.writerow([url])
```

```python
def get_hearing_details(url):
    params = {'api_key': congress_auth.API_KEY}
    r = requests.get(url=url, params=params)
    return r.json()['hearing']
```

```python
with open('hearing_urls.csv', 'r') as f:
    hearing_urls = csv.reader(f)
    hearing_urls = [row[0] for row in hearing_urls]
```

```python
# TODO: Probably need to do some checkpointing here, since this is a lot of data to get.

# This involves opening our CSV file and figuring out the last URL we got details for, and then starting from there.

with open('hearing_details.csv', 'w') as f:
    writer = csv.writer(f)
    for hearing_url in hearing_urls:
        print(f'Getting details for {hearing_url}')
        details = get_hearing_details(hearing_url)
        text_objects = details['formats']
        for text in text_objects:
            if text['type'] == 'Formatted Text':
                url = text['url']
                r = requests.get(url)
                content = r.content
        writer.writerow([details['title'], details['chamber'], details['dates'], hearing_url, content])
```
