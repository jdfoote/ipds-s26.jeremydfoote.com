### Exercise 2

Plot a boxplot of the tip size by the size of the party.

```python
tips = sns.load_dataset('tips')
tips

ax = sns.boxplot(data = tips, y='tip', hue='size', x = 'size', palette='crest')
sns.move_legend(ax, 'upper left', bbox_to_anchor=(1,1));
```

### Exercise 3

Take your code from Exercise 2, and color the boxplots by the gender of the tipper.

```python

ax = sns.boxplot(data = tips, y='tip', hue='sex', x = 'size', palette='crest')
sns.move_legend(ax, 'upper left', bbox_to_anchor=(1,1));
```
