# Example 3: Getting Wikipedia pages

Wikipedia also has an open API, and I want to use it to show one other tip for using the `requests` library; many APIs will take in a set of parameters, which you can pass as a parameter dictionary.

The documentation for the very extensive API is [here](https://www.mediawiki.org/wiki/API:Main_page). Many of the operations require you to authenticate (which we will cover next), but some things, like getting the content of a page, do not.

For example, the following code gets the recent changes to Wikipedia.

````python
import requests

endpt = 'https://en.wikipedia.org/w/api.php'

# Create a session and set a custom user agent. This is important because Wikipedia blocks requests that don't have a user agent, and it's good practice to identify yourself when making requests to an API.
session = requests.Session()
session.headers.update({'User-Agent': 'IPDSTeachingDemo/0.1 (jdfoote@purdue.edu)'})

def get_last_pages_changed(n):
    params = {'action': 'query',
          'format': 'json',
          'list': 'recentchanges',
          'rcnamespace': '0',
          'rclimit': n}
    r = session.get(endpt, params=params)
    #print(r)
    #print(r.json())
    #print(r.json()['query']['recentchanges'])
    result = []
    content = r.json()['query']['recentchanges']
    for page in content:
        result.append(page['title'])
    return result
````

````python

get_last_pages_changed(n = 20)
````

# Example 4: Intro to Bluesky API

Bluesky is a fairly new microblogging platform, which is buillt on the AT Protocol. In theory, this is designed to allow it to be more open an interoperable than other platforms.

One big advantage for researchers is that this means that the data is much more accessible. The `atproto` library allows you to easily access data, and I'll show you how to do a simple search for posts about Purdue, and then how to get the next page of results.

Then, I'll show how to build a follower network for the top posters about Purdue.

First, we'll log in.

If you don't have a Bluesky account, you will need to create one.

Then, in this directory, create a `bluesky_auth.py` file which looks like this:

```
USERNAME = 'YOURUSERNAME'
PASSWORD = 'YOURPASSWORD'
```
In the code below, we'll just authenticate our session, and create a session object (`at_client`), which we can use to run queries.

````python
from atproto import Client
import bluesky_auth

at_client = Client()
at_client.login(login=bluesky_auth.USERNAME, password=bluesky_auth.PASSWORD);
````

Next, we'll do a search for Purdue, and look at what the response looks like.

````python
response = at_client.app.bsky.feed.search_posts({'q':'Purdue', 'limit':100})
print(response)
````

It's this big, ugly Response object. You should play around with it and see if you can figure out how to get the info you want. It took me a while, but I finally figured out that it looks like this:

There's the top level Response, which includes posts, cursor, and hits_total. `posts` is a list of posts, `cursor` is the number you're on in the list of posts, and `hits_total` is None, but I assume it's meant to be the number of results.

Within the `posts` are post objects, which themselves are pretty complicated. In our case, let's just try to get the author, the ID, and the post text.

This is also outlined [in the documentation](https://atproto.blue/en/latest/atproto/atproto_client.models.app.bsky.feed.search_posts.html#module-atproto_client.models.app.bsky.feed.search_posts).

````python
# Get the authors and posts in a list.

posts = []
for post in response.posts:
    author = post.author.handle
    text = post.record.text
    like_count = post.like_count
    repost_count = post.repost_count
    uri = post.uri
    posts.append({'author': author,
                  'text': text,
                  'like_count': like_count,
                  'repost_count': repost_count,
                  'uri': uri                                    
                  })
    
posts
````

Next, let's get multiple pages of results - let's say the last 5000 posts about Purdue. We can do this by using the `cursor` value, which is the number of the last post we got. We can use this to get the next page of results.

````python
def get_posts(query, cursor):
    params = {'q': query, 'limit': 100}
    if cursor is not None:
        params['cursor'] = cursor
    response = at_client.app.bsky.feed.search_posts(params)
    return response


posts = []
cursor = None
while len(posts) < 5000:
    response = get_posts('Purdue', cursor=response.cursor)
    cursor = response.cursor
    for post in response.posts:
        author = post.author.handle
        text = post.record.text
        like_count = post.like_count
        repost_count = post.repost_count
        uri = post.uri
        posts.append({'author': author,
                     'text': text,
                     'like_count': like_count,
                     'repost_count': repost_count,
                     'uri': uri                                    
                     })
    print(f"Retrieved {len(posts)} posts so far...")
````

````python
def get_top_posters(posts):
    poster_counts = {}
    for post in posts:
        author = post['author']
        if author not in poster_counts:
            poster_counts[author] = 0
        poster_counts[author] += 1
    sorted_posters = sorted(poster_counts.items(), key=lambda x: x[1], reverse=True)
    return sorted_posters

top_posters = get_top_posters(posts)
print(top_posters[:10])
````

Finally, let's build a follower network for the top posters about Purdue. To do this, we will need to get who each of the top 100 posters follows, and then see which posters follow each other.

````python
def get_all_follows(client, handle):
    # Get all of the accounts the user follows, handling pagination
    print(f"Getting follows for {handle}...")
    follows = []
    cursor = None
    while True:
        response = client.get_follows(handle, limit=100, cursor=cursor)
        follows.extend(response.follows)
        if response.cursor is None:
            break
        cursor = response.cursor
    return follows

# Get all of the accounts the top posters follow, and see which of the top posters follow each other.
following_dict = {}
top_n = 100
for poster in top_posters[:top_n]:
    name = poster[0]
    follows = get_all_follows(at_client, name)
    following_dict[name] = [f.handle for f in follows]
````

````python
def make_network_from_following(following_dict):
    """
    Take a dictionary of users and their following lists and create a graph.
    The graph is a list of tuples, where each tuple is (user, followed_user).
    """
    following_graph = []
    for user, following in following_dict.items():
        for follow in following:
            if follow in following_dict:
                following_graph.append((user, follow))
    return following_graph

following_graph = make_network_from_following(following_dict)
````

````python
following_graph
````

Finally, let's make a graph of the follower network using `networkx`.

````python
import networkx as nx
import matplotlib.pyplot as plt


G = nx.DiGraph()
G.add_edges_from(following_graph)
# Filter to only reciprocal follows
reciprocal_edges = [(u, v) for u, v in G.edges() if G.has_edge(v, u)]
G_reciprocal = nx.DiGraph()
G_reciprocal.add_edges_from(reciprocal_edges)
plt.figure(figsize=(12, 12))
nx.draw(G_reciprocal, with_labels=True, node_size=500, font_size=8)
````

## Using PRAW

When using PRAW, we need to authenticate. For more complicated APIs, like the Reddit API, it's important for the server to know who is makin the request, so they know what information to receive. Authentication means proving you are who you say you are.

In PRAW, we create a `Reddit` object, which handles authentication and other things. It's basically creating an authenticated session, similar to when you log into a website.

Here, we'll import the PRAW library and our authentication info from the reddit_authenticaton file. Note that there are some things you can do with the API without authenticating, but the rate limit is much higher if you are authenticated. (100 queries per minute vs 10 queries per minute)

````python
import praw
import reddit_authentication
from prawcore.exceptions import NotFound

# Create an instance called reddit. We'll use this to call the API.
reddit = praw.Reddit(client_id=reddit_authentication.client_id,
                     client_secret=reddit_authentication.client_secret,
                    user_agent = reddit_authentication.user_agent,
                    username = reddit_authentication.username,
                    password = reddit_authentication.password)
````

### A note on rate limits

You may have noticed that that took a long time. That's because we are making a lot of requests to the Reddit API. The Reddit API has a rate limit of 100 requests per minute. If you go over the limit, then Reddit sends you a message asking you to wait. PRAW actually handles this without our intervention, but it does mean that it's hard to tell how long things will take. One approach is to at least print out the commenter name, for example, so you can tell how quickly the queries are running. If they are going too slowly, then you may want to change the limits. For example, we got 100 comments per user in this code, but that may take multiple queries, so reducing that number could speed things up.

When using other libraries, including `requests`, you will often have to write code to handle rate limits yourself.
